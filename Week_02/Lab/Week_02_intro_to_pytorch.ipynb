## Initialising PyTorch tensors

A **tensor** is an array of numbers. The same PyTorch class, `torch.Tensor`, represents a scalar, a vector, a matrix, or an array with more dimensions.

The examples below therefore have the **same Python type** but different **shapes**. `type(...)` reports the class of the object; it does not tell us its shape.

As in Week 1, `f"...{expression}..."` inserts the value of the expression into a string. `print()` with no argument prints a blank line.


````python
# A scalar: one number, with no array dimensions.
# The decimal point in 1. makes this a floating-point number.
scalar = t.tensor(1.)
print(f"Scalar type: {type(scalar)}")
print("Scalar value:")
print(scalar)
print()

# A vector containing three ones.
vector = t.ones(3)
print(f"Vector type: {type(vector)}")
print("Vector value:")
print(vector)
print()

# A matrix with three rows and three columns, filled with zeros.
matrix = t.zeros(3, 3)
print(f"Matrix type: {type(matrix)}")
print("Matrix value:")
print(matrix)
print()

# A 3-dimensional array: think of three 3-by-3 matrices of zeros.
threeDtensor = t.zeros(3, 3, 3)
print(f"3D tensor type: {type(threeDtensor)}")
print("3D tensor value:")
print(threeDtensor)
````

````python
# Some other ways to create tensors.
a = t.arange(7)                 # Integers 0, ..., 6: the endpoint 7 is excluded.
print(a)
b = t.linspace(0, 1, 4)         # Four equally spaced points, including 0 and 1.
print(b)
c = t.eye(4)                    # The 4-by-4 identity matrix.
print(c)
d = t.diag(t.tensor([1, 2, 3, 4]))  # Put these four numbers on a matrix diagonal.
print(d)
e = t.rand(4)                   # Four random numbers, uniform in [0, 1).
print(e)
f = t.randn(4)                  # Four standard normal draws (mean 0, std. dev. 1).
print(f)
# Random values will normally differ when you rerun this cell.
````

````python
# These arithmetic operations act element by element, not by Python list rules.
a = t.arange(4)                 # tensor([0, 1, 2, 3])
print(a + 1)                   # Add 1 to every element.
print(5 * a)                   # Multiply every element by 5.
print(a / 2)                   # Divide every element by 2.
print(a + a)                   # Add corresponding elements.
````

## Tensor shape

The **shape** records the size of each dimension, in order. For a matrix, `(3, 4)` means three rows and four columns. For a tensor of shape `(2, 3, 4)`, we can think of two matrices, each with three rows and four columns.

PyTorch numbers dimensions from **0**. In shape `(2, 3, 4)`, dimension `0` has size `2`, dimension `1` has size `3`, and dimension `2` has size `4`.

`tensor.shape` is an **attribute**, so it has no call parentheses. Its value is displayed as `torch.Size([...])`, a tuple-like sequence of dimension sizes.


````python
# Read the shapes of the tensors defined above.
print(scalar.shape)        # torch.Size([]): no dimensions.
print(vector.shape)        # torch.Size([3]): one dimension, of size 3.
print(matrix.shape)        # torch.Size([3, 3]): two dimensions.
print(threeDtensor.shape)  # torch.Size([3, 3, 3]): three dimensions.
````

````python
# Pass an existing shape to create another tensor of that shape.
zero_vector = t.zeros(vector.shape)
print(zero_vector)
````

````python
# These all contain ONE number, but they do not have the same shape.
print(t.zeros(()).shape)     # A scalar; () is the empty shape.
print(t.zeros(1).shape)      # A length-1 vector.
print(t.zeros(1, 1).shape)   # A 1-by-1 matrix.
print(t.zeros(1, 1, 1).shape)# A 1-by-1-by-1 array.

# Use a scalar for a single number unless a calculation needs extra dimensions.
````

````python
# Three ways to arrange the same number of entries.
print(t.zeros(3))     # Shape (3,): a 1-dimensional vector, not a row/column matrix.
print(t.zeros(3, 1))  # Shape (3, 1): an explicit column vector.
print(t.zeros(1, 3))  # Shape (1, 3): an explicit row vector.
# We will see below how a 1-dimensional vector behaves in different operations.
````

### Tensor broadcasting

Suppose we want to add the same vector to every row of a matrix. **Broadcasting** lets an elementwise operation combine certain tensors of different shapes without writing a Python loop.

The shapes determine how the entries are matched. PyTorch does not guess which dimension we intended to use.


````python
# y has one entry for each COLUMN of X.
# The same vector is added to each of X's three rows.
X = t.ones(3, 4)
y = t.randn(4)

print(X)
print(y)
print(X + y)
````

````python
X = t.ones(3, 4)
y = t.rand(3)

# This does not work. The last sizes are 4 and 3, so they cannot broadcast.
# Having three rows in X does not make PyTorch match y to that dimension.
# print(X + y)  # RuntimeError: incompatible sizes.
````

To understand why the first example works and the second does not, use the **broadcasting rules** from the [PyTorch documentation](https://pytorch.org/docs/stable/notes/broadcasting.html):

1. Align the shapes from **right to left**, treating missing dimensions on the left as size `1`.
2. At each position, the sizes must be equal, or one of them must be `1`.
3. A size-`1` dimension is expanded for the calculation to match the other size.

For example, shapes `(3, 4)` and `(4,)` align as `(3, 4)` and `(1, 4)`, so they are compatible. Shapes `(3, 4)` and `(3,)` align as `(3, 4)` and `(1, 3)`, so they are not compatible.


The size-`1` dimensions do not have to appear together or all on one side. For example, the shapes `(1, 5, 3)` and `(4, 1, 3)` are compatible. The first dimension of the first tensor can expand from `1` to `4`, the second dimension of the second tensor can expand from `1` to `5`, and the last dimensions already agree.


````python
X = t.ones(1, 5, 3)
Y = t.arange(12).reshape(4, 1, 3)

Z = X + Y

print(f"X.shape = {X.shape}")
print(f"Y.shape = {Y.shape}")
print(f"Z.shape = {Z.shape}")
````

The resulting shape is `(4, 5, 3)`. Using zero-based indices, the calculation is

$$Z_{ijk}=X_{0jk}+Y_{i0k}.$$

The `0` appears in the first index of `X` because that dimension has only one entry. It appears in the second index of `Y` for the same reason.


Broadcasting also works when the tensors have different numbers of dimensions. PyTorch still aligns their shapes from the right.


````python
X = t.ones(2, 3, 4)
Y = t.arange(4)

Z = X + Y

print(f"X.shape = {X.shape}")
print(f"Y.shape = {Y.shape}")
print(f"Z.shape = {Z.shape}")
````

Here `Y` has shape `(4,)`. For broadcasting, this aligns with `(2, 3, 4)` as if its shape were `(1, 1, 4)`. The calculation is therefore

$$Z_{ijk}=X_{ijk}+Y_k.$$

The length-4 dimension of `Y` matches the last dimension of `X`.


A vector does not automatically match a dimension just because the lengths are equal. For example, a vector of length `2` does **not** automatically match the first dimension of a tensor with shape `(2, 3, 4)`. Its shape `(2,)` is aligned from the right as `(1, 1, 2)`, which conflicts with the final size `4`.


````python
X = t.ones(2, 3, 4)
Y = t.tensor([10., 20.])

print(f"X.shape = {X.shape}")
print(f"Y.shape = {Y.shape}")

# X + Y  # RuntimeError: the last sizes are 4 and 2.
````

If we want the two entries of `Y` to match the **first** dimension of `X`, we need `Y` to have shape `(2, 1, 1)`.

In tensor indexing, `:` means **keep all entries of this dimension**, while `None` inserts a new dimension of size `1`. Therefore `Y[:, None, None]` keeps the two entries of `Y` and adds two size-`1` dimensions after them.


````python
Y = Y[:, None, None]
Z = X + Y

print(f"Y.shape = {Y.shape}")
print(f"Z.shape = {Z.shape}")
print(Z[0, 0])
print(Z[1, 0])
````

Now the calculation is

$$Z_{ijk}=X_{ijk}+Y_{i00}.$$

The first value of `Y` is added throughout the first `3 \times 4` block of `X`, while the second value is added throughout the second block.


## Additional useful operations

For matrices, distinguish **elementwise multiplication**, `A * B`, from **matrix multiplication**, `A @ B`.

`A.mT` gives the matrix transpose. Like `.shape`, it is accessed without call parentheses. In the printed labels below, `\n` starts a new line.


````python
# The tuple (2, 3) specifies the shape: two rows and three columns.
A = t.randn((2, 3))
print("Matrix")
print(A)
print("\nMatrix transpose")
print(A.mT)           # Shape (3, 2).
print("\nElementwise product")
print(A * A)          # Square each entry; shape (2, 3).
print("\nMatrix product")
print(A @ A.mT)       # (2, 3) @ (3, 2) gives shape (2, 2).
````

````python
# Matrix-vector products: for @, the SIDE of the matrix matters.
N = 2
M = 3
A = t.randn((M, N))
a_vec = t.randn(M)  # Shape (M,), not an explicit row or column matrix.
b_vec = t.randn(N)  # Shape (N,).

# Vector on the LEFT: behaves as a row vector of length M.
# The result is still 1-dimensional, with shape (N,).
print((a_vec @ A).shape)

# Vector on the RIGHT: behaves as a column vector of length N.
# The result is still 1-dimensional, with shape (M,).
print((A @ b_vec).shape)

# An explicit row vector: insert a size-1 dimension BEFORE the existing one.
a_rowvec = a_vec[None, :]
print(a_rowvec)
print((a_rowvec @ A).shape)  # (1, M) @ (M, N) gives (1, N).

# An explicit column vector: insert the new dimension AFTER the existing one.
b_colvec = b_vec[:, None]
print(b_colvec)
print((A @ b_colvec).shape)  # (M, N) @ (N, 1) gives (M, 1).
````

````python
# Many tensor operations can be called in two ways:
# t.exp(x) calls a function in the imported library;
# x.exp() calls a method on x. These two calls compute the same result.
x = t.rand(3)
print("Exponential")
print(t.exp(x))
print(x.exp())
print("\n(Natural) logarithm")
print(t.log(x))
print(x.log())
print("\nSine")
print(t.sin(x))
print(x.sin())
print("\nCosine")
print(t.cos(x))
print(x.cos())
# All these calculations act element by element.
````

### Reduction operations

Operations such as `mean`, `sum`, `prod`, `std`, and `var` combine multiple entries into an average, sum, product, standard deviation, or variance.

Here we focus on **which dimension is combined** and **which dimensions remain**. In `A.mean(0)`, the argument `0` selects dimension `0`; it does not select entry `0`.


````python
A = t.randn(2, 3, 4)
````

````python
# By default, reducing one dimension removes that dimension from the shape.
print(A.mean(0).shape)   # (3, 4): average over the size-2 dimension.
print(A.mean(1).shape)   # (2, 4): average over the size-3 dimension.
print(A.mean(2).shape)   # (2, 3): average over the size-4 dimension.

# Negative dimension numbers count from the end.
print(A.mean(-1).shape)  # Same as A.mean(2) for this 3-dimensional tensor.
````

Using zero-based indices to match Python:

`A.mean(0)` averages over `i`, leaving `j` and `k`:

$$M_{jk} = \frac{1}{2}\sum_{i=0}^{1} A_{ijk}.$$

`A.mean(1)` averages over `j`, leaving `i` and `k`:

$$M_{ik} = \frac{1}{3}\sum_{j=0}^{2} A_{ijk}.$$

`A.mean(2)` averages over `k`, leaving `i` and `j`:

$$M_{ij} = \frac{1}{4}\sum_{k=0}^{3} A_{ijk}.$$

Without a dimension argument, `A.mean()` averages **all** entries and returns a scalar tensor. The same dimension-selection idea applies to `sum`.


What if we want to subtract the mean from `A`? After averaging over dimension `0`, the shapes already broadcast:


````python
A_mean = A.mean(0)  # Shape (3, 4), aligned as (1, 3, 4) when subtracting.
print(f"A.shape =      {A.shape}")
print(f"A_mean.shape = {A_mean.shape}")
A - A_mean
````

For this `A`, averaging over dimension `1` or `2` removes a dimension that we still need in order to broadcast the mean back correctly.

For example, `A.mean(2)` has shape `(2, 3)`. Aligned from the right, this becomes `(1, 2, 3)`, which cannot broadcast with `(2, 3, 4)`:


````python
A_mean = A.mean(2)
print(f"A.shape =      {A.shape}")
print(f"A_mean.shape = {A_mean.shape}")
# A - A_mean  # Intentional RuntimeError: these shapes cannot broadcast.
````

We can insert a size-1 dimension **where the reduced dimension used to be**. In this case it was the last dimension:


````python
# First take the mean; then insert the last size-1 dimension.
A_mean = A.mean(2)[:, :, None]  # Shape (2, 3, 1).
print(f"A.shape =      {A.shape}")
print(f"A_mean.shape = {A_mean.shape}")
A - A_mean
````

The [mean operation](https://pytorch.org/docs/stable/generated/torch.mean.html) can do this directly with **`keepdim=True`**.

This is a **keyword argument**: `keepdim` names the option, and `True` enables it. The reduced dimension is retained with size `1` instead of being removed. It is the dimension we reduced—not necessarily the last one—that is retained.


````python
A_mean = A.mean(2, keepdim=True)  # Shape (2, 3, 1), ready to broadcast with A.
print(f"A.shape =      {A.shape}")
print(f"A_mean.shape = {A_mean.shape}")
A - A_mean
````
